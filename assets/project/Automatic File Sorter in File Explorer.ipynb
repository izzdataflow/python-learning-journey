# Automatic File Sorter in File Explorer

```python
import os, shutil
```

```python
#folder path
path = r"D:/Finished project/python-learning-journey/"
```

```python
#folder directory
folder_name = os.listdir(path)
```

```python
#list folder u want to create with for loop
folder_names = ['excel files', 'image files', 'text files']

for loop in range(0,3):
    if not os.path.exists(path + folder_names[loop]):
        os.makedirs((path + folder_names[loop]))

#move the ile into folder
for folder in folder_name:
    if ".xlsx" in folder and not os.path.exists(path + "excel files/" + folder):
        shutil.move(path + folder, path + "excel files/" + folder)
    elif ".jpg" in folder and not os.path.exists(path + "image files/" + folder):
        shutil.move(path + folder, path + "image files/" + folder)
    elif ".txt" in folder and not os.path.exists(path + "text files/" + folder):
        shutil.move(path + folder, path + "text files/" + folder)
```
