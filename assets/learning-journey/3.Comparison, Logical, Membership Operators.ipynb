```python
(70 > 10) and (50 > 10)
```

```python
(10 > 50) or (50 > 10)
```

# Membership Operators

```python
ice_cream = 'I love chocolate ice cream'
'love' in ice_cream
```

```python
scoops = [1,2,3,4,5]
wanted_scoops = 8

wanted_scoops not in scoops
```
