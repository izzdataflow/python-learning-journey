# Converting Data Types

```python
num_int = 7
num_str = '7'
type(num_int)
type(num_str)
```

```python
#not dooable -> different data types
num_sum = num_int + num_str
```

```python
#converting data types
num_str_conv = int(num_str)

#doable
num_sum = num_int + num_str_conv
print(num_sum)
type(num_sum)
```
