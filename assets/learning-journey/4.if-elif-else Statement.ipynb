# if - elif - else Statement

```python
# You can have lot of elif statment but only 1 if and else statement
if 25 > 10:
    print('It worked!')
```

```python
if 25 < 10:
    print('It worked!')
else:
    print('It not work...')
```

```python
if (25 < 10) or (1 < 3):
    print('It worked!')
elif 25 < 20:
    print ('elif 1 worked')
elif 25 < 21:
    print ('elif 2 worked')
elif 25 < 40:
    print ('elif 3 worked')
elif 25 < 50:
    print ('elif 4 worked')
else:
    print('It not worked...')
```

```python
# Nested if statement
if (25 < 10) or (1 < 3):
    print('It worked!')
    if 10 > 5:
        print('This nested if statement worked!')
elif 25 < 20:
    print ('elif 1 worked')
elif 25 < 21:
    print ('elif 2 worked')
elif 25 < 40:
    print ('elif 3 worked')
elif 25 < 50:
    print ('elif 4 worked')
else:
    print('It not worked...')
```
