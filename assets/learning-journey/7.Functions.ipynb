# Function

```python
def first_function():
    print('we did it!')
first_function()
```

```python
#arguments in function
def number_squared(number):
    print(number**2)
number_squared(3)
```

```python
#multiple arguments
def number_squared_cust(number,power):
    print(number**power)
number_squared_cust(5,2)
```

```python
#arbitrary arguments
args_tuple = (5,6,10,12)

def number_args(*number):
    print(number[2]*number[3])
number_args(*args_tuple)
```

```python
#keyword arguments
def number_squared_cust(number,power):
    print(number**power)
number_squared_cust(power = 5, number = 2)
```

```python
#arbitrary keyword arguments
def number_kwargs(**number):
    print('My number is: ' + number['integer'] + 'My other number is: ' + number['integer2'])
number_kwargs(integer = '2309', integer2 = '349')
```
