```python
#lists (can store multiple separate values) (change-able) -> []
[1,2,3]
['Cookie Dough', 'Strawberry', 'Chocolate']
['Vanilla', 3, ['Scoops', 'Spoon'], True]
```

```python
#append , changing lists with index
ice_cream = ['Cookie Dough', 'Strawberry', 'Chocolate']
ice_cream.append('Salted Caramel')
ice_cream[0] = 'Butter Pecan'
ice_cream
```

```python
#tuples (not change-able or immutable after it create) -> ()
tuple_scoops = (1,2,3,2,1)
type(tuple_scoops)
tuple_scoops[0]
```

```python
#set (unordered so can't use index on it) (when ut print only show unique values) -> {}
daily_pints = {1,2,3,1,2,3,8,9,10}
type(daily_pints)
print(daily_pints)
```

```python
wife_daily_pints = {1,2,3,4,5,6}
#finding unique values
print(daily_pints | wife_daily_pints)
```

```python
#finding matches values from both set
print(daily_pints & wife_daily_pints)
```

```python
#finding non-matches values from left set to right set
print(wife_daily_pints - daily_pints)
```

```python
#finding unique values that dont match from both set
print(daily_pints ^ wife_daily_pints)
```

```python
#dictionaries -> {}
#Keys/Values pair

dict_icecream = {'name' : 'Alex Freberg', 'weekly intake' : 5, 'favorite ice creams' : ['MCC', 'Chocolate']}
type(dict_icecream)
print(dict_icecream)
```

```python
dict_icecream.values()
```

```python
dict_icecream.keys()
```

```python
#calling dictionaries is not using index but with key
dict_icecream['name']
```

```python
dict_icecream['name'] = 'Christine Freberg'
```

```python
dict_icecream.update({'name' : 'Christine Freberg', 'weekly intake' : 10, 'weight' : 300})
print(dict_icecream)
```
